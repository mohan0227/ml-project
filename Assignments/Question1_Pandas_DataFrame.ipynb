The below code imports pandas as "pd" and Data frame is created with the given data

```python
import pandas as pd

student_marks_df = pd.DataFrame({
    "StudentID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Name": ['Alice', 'Bob', 'Charlie', 'Diana', 'Ethan', 'Fiona', 'George', 'Hannah', 'Ivan', 'Julia'],
    "Age": [20, 21, 19, 22, 20, 21, 23, 19, 22, 20],
    "Marks": [88, 75, 92, 65, 78, 85, 70, 95, 60, 80],
    "Grade": ['A', 'B', 'A', 'C', 'B', 'A', 'B', 'A', 'C', 'B']
})
```

The below line of code saves the dataframe to a csv file exculding index

```python
student_marks_df.to_csv("students.csv", index=False)
```

The below line of code loads the csv file into a new dataframe

```python
new_df = pd.read_csv("students.csv")
```

```python
print(new_df.head(3))
```
